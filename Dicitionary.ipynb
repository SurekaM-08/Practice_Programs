```python
# Creating a dictionary for student details
student = {
    "Roll Number": "24BAI154",
    "Name": "Sureka M",
    "Department": "AIML",
    "Marks": [98, 95, 88]
}

# Display the student details
print("Student Details:")
for key, value in student.items():
    print(key, ":", value)

# Adding a new field (City)
student["City"] = "Tiruppur"

# Updating the marks
student["Marks"] = [99, 96, 90]

# Removing the department
student.pop("Department")

# Display all keys
print("\nKeys:", list(student.keys()))

# Display all values
print("Values:", list(student.values()))

# Display all key-value pairs
print("Key-Value Pairs:")
for key, value in student.items():
    print(key, ":", value)
```
