# Notebook 1: Create, Update & Save Variable States

**Objective:** Build a notebook that creates, updates, and modifies variables through a loop, then saves these variable states into checkpoint files.

## Step 1: Import TensorFlow

```python
import tensorflow as tf
import os

print("TensorFlow version:", tf.__version__)
```

## Step 2: Create Variables

We create two `tf.Variable` objects that will be modified inside a loop.

```python
# Create variables
counter = tf.Variable(0.0, name="counter")
weight = tf.Variable(1.0, name="weight")

print("Initial counter:", counter.numpy())
print("Initial weight:", weight.numpy())
```

## Step 3: Set Up Checkpoint

A `tf.train.Checkpoint` object tracks the variables we want to save.

```python
checkpoint_dir = "./checkpoints"
os.makedirs(checkpoint_dir, exist_ok=True)
checkpoint_prefix = os.path.join(checkpoint_dir, "ckpt")

checkpoint = tf.train.Checkpoint(counter=counter, weight=weight)
```

## Step 4: Update Variables Through a Loop & Save Checkpoints

Inside the loop, both variables are modified at every step, and a checkpoint file is saved after each update.

```python
num_steps = 5

for step in range(num_steps):
    # Modify variables
    counter.assign_add(1.0)          # increment counter
    weight.assign(weight * 1.5)      # scale weight

    print(f"Step {step + 1}: counter = {counter.numpy()}, weight = {weight.numpy():.4f}")

    # Save checkpoint at this step
    save_path = checkpoint.save(checkpoint_prefix)
    print(f"  -> Saved checkpoint at: {save_path}\n")
```

## Step 5: Verify Checkpoint Files Were Created

```python
print("Files in checkpoint directory:")
for f in os.listdir(checkpoint_dir):
    print(" -", f)

print("\nLatest checkpoint:", tf.train.latest_checkpoint(checkpoint_dir))
```

## Step 6: Clear Notebook Memory

To completely clear the notebook's memory, restart the kernel/runtime:

- **Jupyter Notebook:** `Kernel` menu -> `Restart Kernel`
- **Google Colab:** `Runtime` menu -> `Restart Runtime`

After restarting, all variables (`counter`, `weight`) are wiped from memory, but the checkpoint files remain saved on disk in the `./checkpoints` folder.

Once restarted, do **not** re-run the cells above. Instead, open and run the separate `restore_checkpoint.ipynb` notebook to reload the saved variable states.
