```python
import pandas as pd

# ==========================================
# PART 1: GROUPBY AND AGGREGATION
# ==========================================

# Sample DataFrame based on IPL Player Stats
data = {
    "Player": [
        "Virat Kohli",
        "Rohit Sharma",
        "Shubman Gill",
        "KL Rahul",
        "MS Dhoni",
        "Ruturaj Gaikwad",
        "Jasprit Bumrah",
        "Mohammed Shami",
        "Yuzvendra Chahal",
        "Rashid Khan",
    ],
    "Team": ["RCB", "MI", "GT", "LSG", "CSK", "CSK", "MI", "GT", "RR", "GT"],
    "Matches": [15, 14, 15, 14, 14, 15, 14, 15, 14, 15],
    "Runs": [741, 417, 890, 520, 350, 590, 25, 18, 15, 90],
    "Wickets": [0, 0, 0, 0, 0, 0, 20, 28, 22, 24],
    "Strike_Rate": [
        154.69,
        150.54,
        157.8,
        135.2,
        161.29,
        147.5,
        125.0,
        90.0,
        75.0,
        145.16,
    ],
    "Average": [61.75, 32.08, 59.33, 43.33, 58.33, 42.14, 6.25, 4.5, 3.75, 15.0],
}

df = pd.DataFrame(data)

print("==========================================")
print("             INITIAL DATA FRAME           ")
print("==========================================")
print(df)
print("\n")

# Example 1: Total Runs Scored by Each Team
print("--- Example 1: Total Runs Scored by Each Team ---")
team_runs = df.groupby("Team")["Runs"].sum()
print(team_runs)
print("\n")

# Example 2: Average Runs by Team
print("--- Example 2: Average Runs by Team ---")
avg_runs = df.groupby("Team")["Runs"].mean()
print(avg_runs)
print("\n")

# Example 3: Multiple Aggregate Functions
print("--- Example 3: Multiple Aggregate Functions ---")
result_agg = df.groupby("Team")["Runs"].agg(
    ["sum", "mean", "max", "min", "count"]
)
print(result_agg)
print("\n")

# Example 4: Aggregating Player Data
print("--- Example 4: Aggregating Player Data ---")
player_agg = df.groupby("Player")["Runs"].agg(["sum", "mean", "max"])
print(player_agg)
print("\n")


# ==========================================
# PART 2: JOINING DATAFRAMES
# ==========================================

print("==========================================")
print("            JOINING DATAFRAMES            ")
print("==========================================")

# Creating DataFrames
players = pd.DataFrame(
    {"PlayerID": [1, 2, 3], "Player": ["Virat", "Rohit", "Dhoni"]}
)

scores = pd.DataFrame({"PlayerID": [1, 2, 3], "Runs": [973, 890, 760]})

print("--- DataFrame 1: Players ---")
print(players)
print("\n--- DataFrame 2: Scores ---")
print(scores)
print("\n")

# Inner Join
print("--- Inner Join ---")
inner_result = pd.merge(players, scores, on="PlayerID", how="inner")
print(inner_result)
print("\n")

# Left Join
print("--- Left Join ---")
left_result = pd.merge(players, scores, on="PlayerID", how="left")
print(left_result)
print("\n")

# Right Join
print("--- Right Join ---")
right_result = pd.merge(players, scores, on="PlayerID", how="right")
print(right_result)
print("\n")

# Outer Join
print("--- Outer Join ---")
outer_result = pd.merge(players, scores, on="PlayerID", how="outer")
print(outer_result)
print("\n")

# Join Using Index
print("--- Join Using Index ---")
players_indexed = players.set_index("PlayerID")
scores_indexed = scores.set_index("PlayerID")
index_join_result = players_indexed.join(scores_indexed)
print(index_join_result)
```
