# Notebook 2: Restore Variable States from Checkpoint

**Objective:** After clearing the notebook memory completely, load the previously saved variable states back in from the checkpoint files.

Run this notebook in a **fresh kernel/runtime** (make sure `counter` and `weight` do not already exist in memory).

## Step 1: Import TensorFlow

```python
import tensorflow as tf
import os

print("TensorFlow version:", tf.__version__)
```

## Step 2: Recreate Empty Variables

We must recreate variables with the **same names and structure** as before so the checkpoint knows what to restore into. Their values start fresh/empty (0.0).

```python
counter = tf.Variable(0.0, name="counter")
weight = tf.Variable(0.0, name="weight")

print("Before restore -> counter:", counter.numpy(), " weight:", weight.numpy())
```

## Step 3: Load the Checkpoint

```python
checkpoint_dir = "./checkpoints"

checkpoint = tf.train.Checkpoint(counter=counter, weight=weight)

latest = tf.train.latest_checkpoint(checkpoint_dir)
print("Latest checkpoint found:", latest)

checkpoint.restore(latest).expect_partial()
```

## Step 4: Verify Restored Values

These should match the last printed values from Notebook 1 (before the memory was cleared).

```python
print("After restore -> counter:", counter.numpy(), " weight:", weight.numpy())
```

## Summary

| Stage | counter | weight |
|---|---|---|
| Before restore (fresh vars) | 0.0 | 0.0 |
| After restore (from checkpoint) | matches last saved step | matches last saved step |

This confirms the variable states were successfully saved to disk and reloaded into a completely fresh notebook session.
