```python
import pandas as pd

# Sample student data
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Sureka", "Ranjana", "Srikumaran", "Viji", "Mohan"],
    "Department": ["AI & ML", "CS", "AI & ML", "IT", "CS"],
    "Marks": [85, 76, 92, 68, 81]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print("Student Records:")
print(df)

# Add a new column Grade based on Marks
def assign_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(assign_grade)
print("\nDataFrame with Grade column:")
print(df)

# Display students who scored more than 80 marks
print("\nStudents who scored more than 80 marks:")
print(df[df["Marks"] > 80])

# Calculate the average marks of all students
average_marks = df["Marks"].mean()
print(f"\nAverage marks of all students: {average_marks:.2f}")
```

```python
import pandas as pd

# Sample employee data
data = {
    "Employee_ID": [201, 202, 203, 204, 205],
    "Name": ["Reka", "Ranj", "Kavya", "Tanya", "Priya"],
    "Department": ["HR", "Finance", "IT", "Sales", "IT"],
    "Salary": [45000, 52000, 61000, 38000, 55000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display employee details
print("Employee Records:")
print(df)

# Increase each employee's salary by 10%
df["Salary"] = df["Salary"] * 1.10
print("\nEmployee details after 10% salary increase:")
print(df)

# Display employees whose salary exceeds 50000
print("\nEmployees with salary exceeding Rs. 50000:")
print(df[df["Salary"] > 50000])

# Find the average salary
average_salary = df["Salary"].mean()
print(f"\nAverage salary of employees: Rs. {average_salary:.2f}")
```

```python
import pandas as pd

# Sample sales data
data = {
    "Product": ["Shampoo", "Soap", "Toothpaste", "Face Wash", "Lotion"],
    "Quantity_Sold": [120, 200, 150, 80, 60],
    "Price": [95, 30, 55, 110, 140]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate Total Sales = Quantity x Price
df["Total_Sales"] = df["Quantity_Sold"] * df["Price"]
print("Sales Report:")
print(df)

# Display products with sales greater than 5000
print("\nProducts with sales greater than Rs. 5000:")
print(df[df["Total_Sales"] > 5000])

# Find the product with the highest sales
top_product = df.loc[df["Total_Sales"].idxmax()]
print(f"\nProduct with the highest sales: {top_product['Product']} (Rs. {top_product['Total_Sales']})")
```
